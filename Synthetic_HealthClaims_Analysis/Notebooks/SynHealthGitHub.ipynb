```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Desktop/data-analytics-portfolio/Synthetic_HealthClaims_Analysis/denial_reasons.csv')

df
```

```python
sns.set_style("whitegrid")    

plt.figure(figsize= (10,6))

sns.barplot(
    data = df,
    y= "Reason_Code",
    x= "denial_count",
    hue= "Reason_Code",
    palette= "crest"
)

plt.title("Top 10 Denial Reasons")
plt.ylabel("Reason for Denial")
plt.xlabel("Number of Denials")
plt.tight_layout()
plt.savefig('Desktop/data-analytics-portfolio/Synthetic_HealthClaims_Analysis/Top_Denial_Reasons.png',dpi=300)
plt.show()
```

```python
cdf = pd.read_csv('Desktop/data-analytics-portfolio/Synthetic_HealthClaims_Analysis/Diagnosis_Denial_Rate.csv')
cdf
```

```python
sns.set_style(style = "whitegrid")

plt.figure(figsize= (10,6))

sns.barplot(
    data = cdf,
    x="Diagnosis_Code",
    y="Denial_Rate",
    hue="Diagnosis_Code",
    palette='crest',
    legend=False,
)

plt.title('Top 10 Denied Diagnoses')
plt.xlabel('ICD-10 Code')
plt.ylabel('% of Claims Denied')
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f"{y:.2f}"))
plt.savefig('Desktop/data-analytics-portfolio/Synthetic_HealthClaims_Analysis/Diagnosis_Denial_Rate.png',dpi = 300)
plt.show()
```
